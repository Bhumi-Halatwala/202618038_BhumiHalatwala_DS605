Difference:

* np.linspace() defines an array by the total number of elements and includes the stop value.
* Example: np.linspace(0, 10, num=5) → [0., 2.5, 5., 7.5, 10.]
* np.arange() defines an array by the step size between numbers and excludes the stop value.
* Example: np.arange(0, 10, step=2) → [0, 2, 4, 6, 8]

## Task 2 - Vectorized Arithmetic and Linear Algebra

#### Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().

```python
# First 2 x 2 matrix
matrix_A = np.array([
    [1, 2],
    [3, 4]
])

# Second 2 x 2 matrix
matrix_B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(matrix_A)

print("\nMatrix B:")
print(matrix_B)
```

```python
# Matrix Addition
addition = matrix_A + matrix_B
print("\nMatrix Addition:")
print(addition)

# Element-wise Multiplication
elementwise_multiplication = matrix_A * matrix_B
print("\nElement-wise Multiplication:")
print(elementwise_multiplication)

# Matrix Multiplication (Dot Product)
matrix_multiplication = matrix_A @ matrix_B
print("\nMatrix Multiplication (Dot Product):")
print(matrix_multiplication)
```

#### For a square matrix, find its transpose and determinant.

```python
transpose_A = matrix_A.T
print("\nTranspose of Matrix A:")
print(transpose_A)

determinant_A = np.linalg.det(matrix_A)
print("\nDeterminant of Matrix A:")
print(determinant_A)
```

#### If invertible, find the inverse and verify A @ A^-1 using np.allclose().

```python
if np.linalg.det(matrix_A) != 0:

    # Calculate the inverse of Matrix A
    inverse_A = np.linalg.inv(matrix_A)

    print("Inverse of Matrix A:")
    print(inverse_A)

    verification = matrix_A @ inverse_A

    print("\nA @ A^-1:")
    print(verification)

    verification = np.round(verification, 10)

    print("\nIdentity Matrix:")
    print(np.eye(matrix_A.shape[0]))

else:
    print("Matrix A is not invertible.")
```

### Task 5 - Filtering and Querying

#### Male passengers older than 50: how many?

```python
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

male_over_50_count = len(male_over_50)

print("Male passengers older than 50:", male_over_50_count)
```

#### Female first-class passengers: how many, and what percentage survived?

```python
female_first_class = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

female_first_class_count = len(female_first_class)

female_first_class_survival = (female_first_class["Survived"].mean() * 100)

print("Female first-class passengers:", female_first_class_count)
print("Percentage survived:", round(female_first_class_survival, 2), "%")
```

#### Age 20-40, Fare above overall median, and survived: how many?

```python
overall_median_fare = df["Fare"].median()

age_fare_survived = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]

age_fare_survived_count = len(age_fare_survived)

print("Overall median Fare:", round(overall_median_fare, 2))
print("Passengers aged 20-40, above median Fare, and survived:", age_fare_survived_count)
```

#### Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?

```python
alone_under_30_died = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

alone_under_30_died_count = len(alone_under_30_died)

print("Passengers travelling alone, age < 30, and did not survive:", alone_under_30_died_count)
```

#### Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python
southampton_median_fare = df.loc[df["Embarked"] == "S", "Fare"].median()

southampton_result = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

southampton_count = len(southampton_result)

print("Southampton median Fare:", round(southampton_median_fare, 2))
print("Passengers satisfying the conditions:", southampton_count)
```

### Task 6 - groupby() and Aggregation

#### Survival rate by Sex.

```python
survival_by_sex = (
    df.groupby("Sex")["Survived"]
    .mean()
    .mul(100)
    .round(2)
)

print("Survival Rate by Sex:")
display(survival_by_sex)
```

#### Survival rate by Pclass.

```python
survival_by_pclass = (
    df.groupby("Pclass")["Survived"]
    .mean()
    .mul(100)
    .round(2)
)

print("Survival Rate by Pclass:")
display(survival_by_pclass)
```

#### Average Age and Fare by Pclass.

```python
average_by_pclass = (
    df.groupby("Pclass")[["Age", "Fare"]]
    .mean()
    .round(2)
)

print("Average Age and Fare by Pclass:")
display(average_by_pclass)
```

#### Passenger count and survival rate by Sex-Pclass.

```python
sex_pclass_summary = (
    df.groupby(["Sex", "Pclass"])
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Survival_Rate=("Survived", "mean")
    )
)

sex_pclass_summary["Survival_Rate"] = (sex_pclass_summary["Survival_Rate"] * 100).round(2)

print("Passenger Count and Survival Rate by Sex-Pclass:")
display(sex_pclass_summary)
```

#### Passenger count, average Fare, and survival rate by Embarked.

```python
embarked_summary = (
    df.groupby("Embarked")
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Average_Fare=("Fare", "mean"),
        Survival_Rate=("Survived", "mean")
    )
)

embarked_summary["Average_Fare"] = (embarked_summary["Average_Fare"].round(2))

embarked_summary["Survival_Rate"] = (embarked_summary["Survival_Rate"] * 100).round(2)

print("Summary by Embarked:")
display(embarked_summary)
```

### Task 7 - Missing Values and Fare Outliers

#### Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.

#### Fill missing Age values with the mean Age. Show missing counts before and after.

```python
df_mean_age = df.copy()

missing_age_before = df_mean_age["Age"].isnull().sum()

age_mean = df_mean_age["Age"].mean()

df_mean_age["Age"] = df_mean_age["Age"].fillna(age_mean)

missing_age_after = df_mean_age["Age"].isnull().sum()

print("Missing Age before imputation:", missing_age_before)
print("Mean Age used for imputation:", round(age_mean, 2))
print("Missing Age after imputation:", missing_age_after)
```

#### Try to fill NaN values by imputing using: mean, median, mode, random value

```python
df_age_mean = df.copy()

df_age_mean["Age"] = df_age_mean["Age"].fillna(df_age_mean["Age"].mean())

print("Missing Age after mean imputation:", df_age_mean["Age"].isnull().sum())
```

```python
df_age_median = df.copy()

df_age_median["Age"] = df_age_median["Age"].fillna(df_age_median["Age"].median())

print("Missing Age after median imputation:", df_age_median["Age"].isnull().sum())
```

```python
df_age_mode = df.copy()

age_mode = df_age_mode["Age"].mode()[0]

df_age_mode["Age"] = df_age_mode["Age"].fillna(age_mode)

print("Mode Age used:", age_mode)

print("Missing Age after mode imputation:", df_age_mode["Age"].isnull().sum())
```

```python
df_age_random = df.copy()

np.random.seed(42)

missing_age_mask = df_age_random["Age"].isnull()

random_age_values = np.random.choice(
    df_age_random["Age"].dropna(),
    size=missing_age_mask.sum()
)

df_age_random.loc[
    missing_age_mask,
    "Age"
] = random_age_values

print("Missing Age after random imputation:", df_age_random["Age"].isnull().sum())
```

#### For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers.

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

outlier_count = len(fare_outliers)

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))
print("Lower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))
print("Number of Fare Outliers:", outlier_count)
```

### Task 8 — Features and Pivot Table

#### Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.

```python
df["FamilySize"] = (df["SibSp"] + df["Parch"] + 1)

df["IsAlone"] = np.where(df["FamilySize"] == 1, 1, 0)

display(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10))
```

#### Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest groups.

```python
survival_pivot = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print("Survival Rate Pivot Table:")
display(survival_pivot)
```

```python
highest_group = survival_pivot.stack().idxmax()
highest_rate = survival_pivot.stack().max()

lowest_group = survival_pivot.stack().idxmin()
lowest_rate = survival_pivot.stack().min()

print("Highest survival group:", highest_group)

print("Highest survival rate:", round(highest_rate * 100, 2), "%")

print("\nLowest survival group:", lowest_group)

print("Lowest survival rate:", round(lowest_rate * 100, 2), "%")
```

### Task 9 - Visualizations and Observations

#### Plot a correlation heatmap for relevant numerical columns. Note the strongest positive and negative relationships.

```python
numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation_matrix = df[numerical_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Create a copy of the correlation matrix
corr_values = correlation_matrix.copy()

# Keep only the upper triangle and remove the diagonal
upper_triangle = np.triu(np.ones(corr_values.shape, dtype=bool), k=1)

# Convert the selected correlations into variable pairs
corr_pairs = corr_values.where(upper_triangle).stack()

# Identify strongest positive relationship
strongest_positive_pair = corr_pairs.idxmax()
strongest_positive_value = corr_pairs.max()

# Identify strongest negative relationship
strongest_negative_pair = corr_pairs.idxmin()
strongest_negative_value = corr_pairs.min()

# Display results
print("Strongest Positive Relationship:")
print(
    f"{strongest_positive_pair[0]} and "
    f"{strongest_positive_pair[1]}: "
    f"{strongest_positive_value:.2f}"
)

print("\nStrongest Negative Relationship:")
print(
    f"{strongest_negative_pair[0]} and "
    f"{strongest_negative_pair[1]}: "
    f"{strongest_negative_value:.2f}"
)
```

#### Saving Cleaned Dataset

```python
# Create a copy of the original dataset for cleaning
cleaned_df = df.copy()

# Fill missing Age values using the mean Age
cleaned_df["Age"] = cleaned_df["Age"].fillna(cleaned_df["Age"].mean())

# Fill missing Embarked values using the mode
cleaned_df["Embarked"] = cleaned_df["Embarked"].fillna(cleaned_df["Embarked"].mode()[0])

# Fill missing Fare values using the median Fare
cleaned_df["Fare"] = cleaned_df["Fare"].fillna(cleaned_df["Fare"].median())

# Fill missing Cabin values using the mode Cabin
cleaned_df["Cabin"] = cleaned_df["Fare"].fillna(cleaned_df["Cabin"].mode())

# Save the cleaned dataset as a CSV file
cleaned_df.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned Titanic dataset saved as titanic_cleaned.csv")

print("\nRemaining missing values:")
print(cleaned_df.isnull().sum())
```
